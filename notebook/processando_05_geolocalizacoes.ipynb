```python
from project_olist.config import RAW_DATA_DIR
import pandas

dataframe_geolocalizacoes = pandas.read_csv(RAW_DATA_DIR / "olist_geolocation_dataset.csv")
dataframe_geolocalizacoes = dataframe_geolocalizacoes.rename(
    columns={
        "geolocation_zip_code_prefix": "prefixo_cep",
        "geolocation_lat": "latitude",
        "geolocation_lng": "longitude",
        "geolocation_city": "cidade",
        "geolocation_state": "estado"
    }
)

dataframe_geolocalizacoes.info()
```

```python
dataframe_geolocalizacoes["prefixo_cep"] = dataframe_geolocalizacoes["prefixo_cep"].astype(str)
dataframe_geolocalizacoes["prefixo_cep"] = dataframe_geolocalizacoes["prefixo_cep"].map(
    lambda x: x.rjust(5, "0")
)
```

```python
dataframe_geolocalizacoes.sample(n=5)
```

```python
dataframe_geolocalizacoes.describe()
```

```python
dataframe_geolocalizacoes = dataframe_geolocalizacoes.groupby(by="prefixo_cep", as_index=False).agg(
    {
        "latitude": "mean",
        "longitude": "mean",
        "cidade": "first",
        "estado": "first"
    }
)
```

```python
dataframe_geolocalizacoes.describe()
```

```python
from project_olist.config import PROCESSED_DATA_DIR

dataframe_geolocalizacoes.to_csv(PROCESSED_DATA_DIR / "geolocalizacoes.csv", index=False)
```
