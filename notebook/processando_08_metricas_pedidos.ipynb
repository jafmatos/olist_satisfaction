```python
from project_olist.config import PROCESSED_DATA_DIR
import pandas

dataframe_avaliacoes = pandas.read_csv(PROCESSED_DATA_DIR / "avaliacoes.csv")
dataframe_pedidos = pandas.read_csv(PROCESSED_DATA_DIR / "pedidos.csv")
dataframe_itens_pedidos = pandas.read_csv(PROCESSED_DATA_DIR / "itens_pedidos.csv") 
dataframe_metricas_produtos = pandas.read_csv(PROCESSED_DATA_DIR / "produtos_metricas.csv")
```

```python
dataframe_metricas_produtos.info()
```

```python
dataframe_pedidos = dataframe_pedidos.merge(
    dataframe_avaliacoes,
    on="id_pedido",
    how="left"
)

dataframe_pedidos.info()
```

```python
dataframe_pedidos["pontuacao"] = (
    dataframe_pedidos["pontuacao"].fillna(
        dataframe_pedidos["pontuacao"].median()
    )
)

dataframe_pedidos["tempo_resposta_dias"] = (
    dataframe_pedidos["tempo_resposta_dias"].fillna(
        dataframe_pedidos["tempo_resposta_dias"].median()
    )
)

dataframe_pedidos["cliente_satisfeito"] = (
    dataframe_pedidos["cliente_satisfeito"].fillna(
        dataframe_pedidos["cliente_satisfeito"].median()
    )
)

dataframe_pedidos.info()
```

```python
dataframe_pedidos = dataframe_pedidos.rename(
    columns={
        "pontuacao": "pontuacao_avaliacao",
        "tempo_resposta_dias": "tempo_resposta_dias_avaliacao"
    }
)

dataframe_pedidos.info()
```

```python
dataframe_metricas_pedidos = dataframe_pedidos.copy()

dataframe_metricas_pedidos = dataframe_metricas_pedidos.merge(
    dataframe_itens_pedidos,
    on="id_pedido",
    how="inner"
)

dataframe_metricas_pedidos.info()
```

```python
dataframe_metricas_pedidos = dataframe_metricas_pedidos.merge(
    dataframe_metricas_produtos,
    on="id_produto",
    how="left"
)

dataframe_metricas_pedidos.info()
```

```python
dataframe_metricas_pedidos = dataframe_metricas_pedidos.groupby(by="id_pedido", as_index=False).agg(
    {
        "preco": "sum",
        "frete": "sum",
        "volume_vendas": "mean",
        "media_avaliacao": "mean"
    }
)

dataframe_metricas_pedidos = dataframe_metricas_pedidos.rename(
    columns={
        "preco": "preco_total_itens",
        "frete": "preco_total_frete",
        "volume_vendas": "media_volume_vendas_produtos",
        "media_avaliacao": "media_avaliacao_produtos"
    }
)

dataframe_metricas_pedidos.info()
```

```python
dataframe_metricas_pedidos.to_csv(PROCESSED_DATA_DIR / "pedidos_metricas.csv")
```

```python
dataframe_pedidos = dataframe_pedidos.merge(
    dataframe_metricas_pedidos,
    on="id_pedido",
    how="left"
)

dataframe_pedidos.info()
```

```python
dataframe_pedidos.sample(n=5)
```

```python
dataframe_pedidos["media_avaliacao_produtos"] = dataframe_pedidos["media_avaliacao_produtos"].round(2)
dataframe_pedidos["media_volume_vendas_produtos"] = dataframe_pedidos["media_volume_vendas_produtos"].round(2)

dataframe_pedidos.sample(n=5)
```

```python
dataframe_pedidos["preco_total_pedido"] = dataframe_pedidos["preco_total_itens"] + dataframe_pedidos["preco_total_frete"]
```

```python
dataframe_clientes = pandas.read_csv(PROCESSED_DATA_DIR / "clientes.csv")

dataframe_pedidos = dataframe_pedidos.merge(
    dataframe_clientes,
    on="id_cliente",
    how="left"
)

dataframe_pedidos.info()
```

```python
dataframe_geolocalizacoes = pandas.read_csv(PROCESSED_DATA_DIR / "geolocalizacoes.csv")

dataframe_pedidos = dataframe_pedidos.merge(
    dataframe_geolocalizacoes,
    left_on="prefixo_cep",
    right_on="prefixo_cep",
    how="left"
)

dataframe_pedidos.info()
```

```python
dataframe_pedidos = dataframe_pedidos.drop(
    columns=[
        "id_unico_cliente",
        "prefixo_cep",
        "cidade_y",
        "estado_y"
    ]
).rename(
    columns={
        "cidade_x": "cidade",
        "estado_x": "estado"
    }
)

dataframe_pedidos.info()
```
