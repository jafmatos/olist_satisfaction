```python
from project_olist.config import RAW_DATA_DIR
import pandas

dataframe_clientes = pandas.read_csv(RAW_DATA_DIR / "olist_customers_dataset.csv")
dataframe_clientes = dataframe_clientes.rename(
    columns={
        "customer_id": "id_cliente",
        "customer_unique_id": "id_unico_cliente",
        "customer_zip_code_prefix": "prefixo_cep",
        "customer_city": "cidade",
        "customer_state": "estado"
    }
)

dataframe_clientes.info()
```

```python
dataframe_clientes["prefixo_cep"] = dataframe_clientes["prefixo_cep"].astype(str)
dataframe_clientes["prefixo_cep"] = dataframe_clientes["prefixo_cep"].map(
    lambda x: x.rjust(5, "0")
)
```
