```python
from project_olist.config import PROCESSED_DATA_DIR
import pandas

dataframe_avaliacoes = pandas.read_csv(PROCESSED_DATA_DIR / "avaliacoes.csv")
dataframe_itens_pedidos = pandas.read_csv(PROCESSED_DATA_DIR / "itens_pedidos.csv") 
dataframe_produtos = pandas.read_csv(PROCESSED_DATA_DIR / "produtos.csv")
```

```python
dataframe_itens_pedidos.info()
```

```python
dataframe_produtos.info()
```

```python
dataframe_produtos = dataframe_produtos.merge(
    dataframe_itens_pedidos,
    on="id_produto",
    how="left"
)

dataframe_produtos.info()
```

```python
dataframe_produtos = dataframe_produtos.merge(
    dataframe_avaliacoes,
    on="id_pedido",
    how="left"
)

dataframe_produtos.info()
```

```python
dataframe_metricas_produtos = dataframe_produtos.copy()
dataframe_metricas_produtos = dataframe_metricas_produtos.groupby(by="id_produto", as_index=False).agg(
    {
        "id_pedido": "count",
        "pontuacao": "mean",
    }
)

dataframe_metricas_produtos = dataframe_metricas_produtos.rename(
    columns={
        "id_pedido": "volume_vendas",
        "pontuacao": "media_avaliacao"
    }
)

dataframe_metricas_produtos.sample(n=5)
```

```python
dataframe_metricas_produtos.info()
```

```python
dataframe_metricas_produtos["media_avaliacao"] = (
    dataframe_metricas_produtos["media_avaliacao"].fillna(
        dataframe_metricas_produtos["media_avaliacao"].median()
    )
)
```
