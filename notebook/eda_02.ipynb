```python
from project_olist.config import RAW_DATA_DIR
import pandas

dataframe_avaliacoes = pandas.read_csv(RAW_DATA_DIR / "olist_order_reviews_dataset.csv")
dataframe_avaliacoes = dataframe_avaliacoes.rename(
    columns={
        "review_id": "id_avaliacao",
        "order_id": "id_pedido",
        "review_score": "pontuacao",
        "review_creation_date": "data_criacao",
        "review_answer_timestamp": "data_resposta"
    }
)

dataframe_avaliacoes = dataframe_avaliacoes.drop(
    columns=["review_comment_title", "review_comment_message"]
)

dataframe_avaliacoes.info()
```

```python
dataframe_avaliacoes.sample(n=5)
```

```python
dataframe_avaliacoes["data_criacao"] = pandas.to_datetime(dataframe_avaliacoes["data_criacao"]).dt.normalize()
dataframe_avaliacoes["data_resposta"] = pandas.to_datetime(dataframe_avaliacoes["data_resposta"]).dt.normalize()

dataframe_avaliacoes.info()
```

```python
dataframe_avaliacoes["tempo_resposta_dias"] = (
    (dataframe_avaliacoes["data_resposta"] - dataframe_avaliacoes["data_criacao"]).dt.days
)

dataframe_avaliacoes.sample(n=5)
```

```python
dataframe_avaliacoes = dataframe_avaliacoes.drop(
    columns=[
        "data_criacao",
        "data_resposta"
    ]
)

dataframe_avaliacoes.info()
```

```python
dataframe_avaliacoes.describe()
```

```python
dataframe_avaliacoes["cliente_satisfeito"] = dataframe_avaliacoes.apply(
    lambda row: (
        1 if row["pontuacao"] >= 4
        else 0
    ),
    axis=1
)
```

```python
from project_olist.config import PROCESSED_DATA_DIR

dataframe_avaliacoes.to_csv(PROCESSED_DATA_DIR / "avaliacoes.csv", index=False)
```
