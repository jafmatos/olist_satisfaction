```python
from project_olist.config import RAW_DATA_DIR
import pandas

dataframe_produtos = pandas.read_csv(RAW_DATA_DIR / "olist_products_dataset.csv")
dataframe_produtos = dataframe_produtos.rename(
    columns={
        "product_id": "id_produto",
        "product_category_name": "categoria_produto",
        "product_name_lenght": "tamanho_nome_produto",
        "product_description_lenght": "tamanho_descricao_produto",
        "product_photos_qty": "quantidade_imagens_produto",
        "product_weight_g": "peso_gramas_produto",
        "product_length_cm": "comprimento_cm_produto",
        "product_height_cm": "altura_cm_produto",
        "product_width_cm": "largura_cm_produto"
    }
)

dataframe_produtos.info()
```

```python
dataframe_produtos = dataframe_produtos.drop_duplicates(subset="id_produto")

dataframe_produtos.info()
```

```python
dataframe_produtos = dataframe_produtos[["id_produto", "categoria_produto"]]
```
