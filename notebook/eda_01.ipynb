```python
from project_olist.config import PROCESSED_DATA_DIR
import pandas
import seaborn
import matplotlib.pyplot as pyplot

dataframe = pandas.read_csv(PROCESSED_DATA_DIR / "pedidos_com_metricas.csv")

dataframe.info()
```

```python
dataframe["entrega_atrasada"] = dataframe.apply(
    lambda row: (
        1 if row["dias_para_entrega"] > row["dias_para_entrega_estimada"] 
        else 0
    ),
    axis=1
)

dataframe["atraso_entrega_dias"] = dataframe.apply(
    lambda row: (
        0 if row["entrega_atrasada"] == 0 
        else row["dias_para_entrega"] - row["dias_para_entrega_estimada"]
    ),
    axis=1
)

pyplot.figure(figsize=(10, 10))

seaborn.heatmap(
    data=dataframe.select_dtypes(include="number").corr(),
    cmap="coolwarm",
    annot=True,
    fmt=".2f"
)
```

```python
seaborn.scatterplot(
    data=dataframe,
    x="dias_para_entrega",
    y="dias_para_entrega_estimada",
    hue="pontuacao_avaliacao",
    alpha=0.3
)
```

```python
seaborn.histplot(
    data=dataframe[dataframe["preco_total_frete"] < 400],
    x="preco_total_frete"
)
```

```python
seaborn.scatterplot(
    data=dataframe[dataframe["preco_total_frete"] < 50],
    x="preco_total_itens",
    y="preco_total_frete",
    hue="pontuacao_avaliacao",
    alpha=0.3
)
```

```python
import numpy

dataframe["fator_frete_dias_atraso"] = dataframe["preco_total_frete"] * dataframe["atraso_entrega_dias"]

dataframe["cliente_satisfeito"] = dataframe["pontuacao_avaliacao"].map(
    lambda x: 1 if x >= 4 else 0
)

dataframe.replace([numpy.inf, -numpy.inf], numpy.nan)
```

```python
seaborn.scatterplot(
    data=dataframe,
    x="atraso_entrega_dias",
    y="fator_frete_dias_atraso",
    hue="cliente_satisfeito",
    alpha=0.1,
    size=0.0001
)
```
