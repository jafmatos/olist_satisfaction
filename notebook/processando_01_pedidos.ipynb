```python
from project_olist.config import RAW_DATA_DIR
import pandas

dataframe_pedidos = pandas.read_csv(RAW_DATA_DIR / "olist_orders_dataset.csv")

dataframe_pedidos = dataframe_pedidos.rename(
    columns={
        "order_id": "id_pedido",
        "customer_id": "id_cliente",
        "order_status": "status_pedido",
        "order_purchase_timestamp": "data_criacao",
        "order_approved_at": "data_aprovacao_pagamento",
        "order_delivered_carrier_date": "data_postagem",
        "order_delivered_customer_date": "data_entrega",
        "order_estimated_delivery_date": "data_estimada_entrega"
    }
)


dataframe_pedidos.info()
```

```python
dataframe_pedidos = dataframe_pedidos.drop_duplicates(subset="id_pedido")
dataframe_pedidos = dataframe_pedidos.reset_index(drop=True)

dataframe_pedidos.info()
```

```python
dataframe_pedidos.sample(n=5)
```

```python
dataframe_pedidos["status_pedido"].unique()
```

```python
dataframe_pedidos["status_pedido"] = dataframe_pedidos["status_pedido"].map({
    "delivered": "entregue",
    "invoiced": "faturado",
    "shipped": "enviado",
    "processing": "processando",
    "unavailable": "indisponível",
    "canceled": "cancelado",
    "created": "criado",
    "approved": "aprovado"
})
```

```python
dataframe_pedidos = dataframe_pedidos[dataframe_pedidos["status_pedido"] == "entregue"]
```

```python
dataframe_pedidos["data_criacao"] = pandas.to_datetime(dataframe_pedidos["data_criacao"]).dt.normalize()
dataframe_pedidos["data_aprovacao_pagamento"] = pandas.to_datetime(dataframe_pedidos["data_aprovacao_pagamento"]).dt.normalize()
dataframe_pedidos["data_postagem"] = pandas.to_datetime(dataframe_pedidos["data_postagem"]).dt.normalize()
dataframe_pedidos["data_entrega"] = pandas.to_datetime(dataframe_pedidos["data_entrega"]).dt.normalize()
dataframe_pedidos["data_estimada_entrega"] = pandas.to_datetime(dataframe_pedidos["data_estimada_entrega"]).dt.normalize()

dataframe_pedidos.info()
```

```python
dataframe_pedidos["dias_para_entrega_estimada"] = (
    (dataframe_pedidos["data_estimada_entrega"] - dataframe_pedidos["data_criacao"]).dt.days
)

dataframe_pedidos["dias_para_entrega_estimada"] = (
    dataframe_pedidos["dias_para_entrega_estimada"]
        .fillna(dataframe_pedidos["dias_para_entrega_estimada"].median())
)

dataframe_pedidos["dias_para_entrega_estimada"] = (
    dataframe_pedidos["dias_para_entrega_estimada"].astype("int")
)

dataframe_pedidos["dias_para_postagem"] = (
    (dataframe_pedidos["data_postagem"] - dataframe_pedidos["data_criacao"]).dt.days
)

dataframe_pedidos["dias_para_postagem"] = (
    dataframe_pedidos["dias_para_postagem"]
        .fillna(dataframe_pedidos["dias_para_postagem"].median())
)

dataframe_pedidos["dias_para_postagem"] = (
    dataframe_pedidos["dias_para_postagem"].astype("int")
)

dataframe_pedidos["dias_para_entrega"] = (
    (dataframe_pedidos["data_entrega"] - dataframe_pedidos["data_criacao"]).dt.days
)

dataframe_pedidos["dias_para_entrega"] = (
    dataframe_pedidos["dias_para_entrega"]
        .fillna(dataframe_pedidos["dias_para_entrega"].median())
)

dataframe_pedidos["dias_para_entrega"] = (
    dataframe_pedidos["dias_para_entrega"].astype("int")
)

dataframe_pedidos.sample(n=5)
```

```python
dataframe_pedidos.describe()
```

```python
dataframe_pedidos = dataframe_pedidos[dataframe_pedidos["dias_para_postagem"] >= 0]

dataframe_pedidos.describe()
```

```python
dataframe_pedidos = dataframe_pedidos.drop(
    columns=[
        "data_criacao",
        "data_aprovacao_pagamento",
        "data_postagem",
        "data_entrega",
        "data_estimada_entrega"
    ]
)

dataframe_pedidos.info()
```
